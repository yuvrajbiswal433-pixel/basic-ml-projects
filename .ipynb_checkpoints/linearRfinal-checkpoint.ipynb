```python
import numpy as np
import pandas as pd 
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.base import BaseEstimator, TransformerMixin
from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OrdinalEncoder,OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.linear_model import Lasso
from sklearn.metrics import r2_score, mean_squared_error
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import GridSearchCV
```

```python
df=pd.read_csv("hptrain.csv")
df.sample(5)
```

MISSING COLS

```python
missing_percentage = (df.isnull().sum() / len(df)) * 100
missing_percentage = missing_percentage[missing_percentage > 0]
missing_df = pd.DataFrame({
    'Missing_Percentage': missing_percentage,
    'Dtype': df[missing_percentage.index].dtypes
})
print(missing_df.sort_values(by='Missing_Percentage', ascending=False))
```

```python
plt.figure(figsize=(30,30))
sns.heatmap(df.select_dtypes(include=['number']).corr(), cmap='OrRd',annot=True)
```

TRAIN TEST SPLIT

```python
x=df.drop(['SalePrice'],axis=1)
y=df['SalePrice']
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
cols = ['GarageYrBlt','MasVnrArea']
x_train[cols].hist(figsize=(15,8))
```

NUMERIC IMPUTER

```python
cols1 = ['GarageYrBlt',
    'MasVnrArea',
    'BsmtFinSF1',
    'BsmtFinSF2',
    'BsmtUnfSF',
    'TotalBsmtSF',
    'BsmtFullBath',
    'BsmtHalfBath',
    'GarageCars',
    'GarageArea']
class numimputer(BaseEstimator,TransformerMixin):
    def fit(self,x,y=None):
        return self
    def transform(self,x):
        si=SimpleImputer(strategy="median")
        x[cols1]=si.fit_transform(x[cols1])
        return x
```

DROPPER

```python
cols2=['PoolQC','MiscFeature','Alley','Fence','MasVnrType','FireplaceQu','LotFrontage','Id','PoolArea','3SsnPorch','MiscVal','LowQualFinSF']
class dropper(BaseEstimator,TransformerMixin):
    def fit(self,x,y=None):
        return self
    def transform(self,x):
        x=x.drop(cols2,axis=1)
        return x
```

ORDINAL ENCODING

```python
# basement orders
bsmt_qual_order = ['Missing','Po', 'Fa', 'TA', 'Gd', 'Ex']
bsmt_cond_order = ['Missing', 'Po', 'Fa', 'TA', 'Gd', 'Ex']
bsmt_exposure_order = ['Missing', 'No', 'Mn', 'Av', 'Gd']
bsmtfintype_order = [
    'Missing',
    'Unf',
    'LwQ',
    'Rec',
    'BLQ',
    'ALQ',
    'GLQ'
]
# garage orders
garage_finish_order = [
    'Missing',
    'Unf',
    'RFn',
    'Fin'
]
garage_qual_cond_order = [
    'Missing',
    'Po',
    'Fa',
    'TA',
    'Gd',
    'Ex'
]
# exterior quality orders
exter_order = [
    'Po',
    'Fa',
    'TA',
    'Gd',
    'Ex'
]
# heating quality
heatingqc_order = [
    'Po',
    'Fa',
    'TA',
    'Gd',
    'Ex'
]
# kitchen quality
kitchenqual_order = [
    'Missing',
    'Po',
    'Fa',
    'TA',
    'Gd',
    'Ex'
]
# all ordinal columns
cols3 = [
    'ExterQual',
    'ExterCond',
    'BsmtQual',
    'BsmtCond',
    'BsmtExposure',
    'BsmtFinType1',
    'BsmtFinType2',
    'HeatingQC',
    'KitchenQual',
    'GarageFinish',
    'GarageQual',
    'GarageCond'
]
class ordinalencoder(BaseEstimator, TransformerMixin):
    def fit(self, x, y=None):
        self.oe = OrdinalEncoder(
            categories=[
                exter_order,
                exter_order,
                bsmt_qual_order,
                bsmt_cond_order,
                bsmt_exposure_order,
                bsmtfintype_order,
                bsmtfintype_order,
                heatingqc_order,
                kitchenqual_order,
                garage_finish_order,
                garage_qual_cond_order,
                garage_qual_cond_order
            ]
        )
        self.oe.fit(x[cols3])
        return self
    def transform(self, x):
        x[cols3] = self.oe.transform(x[cols3])
        return x        
```

ONE HOT ENCODER

```python
ohe_cols = [
    'MSZoning','Street','LotShape','LandContour','Utilities',
    'LotConfig','LandSlope','Neighborhood','Condition1',
    'Condition2','BldgType','HouseStyle','RoofStyle',
    'RoofMatl','Exterior1st','Exterior2nd','Foundation',
    'Heating','CentralAir','Electrical','Functional',
    'GarageType','PavedDrive','SaleType','SaleCondition'
]

class OHEEncoder(BaseEstimator, TransformerMixin):
    def fit(self, x, y=None):
        self.encoder = OneHotEncoder(
            drop='first',
            sparse_output=False,
            handle_unknown='ignore'
        )
        self.encoder.fit(x[ohe_cols])
        return self
    def transform(self, x):
        encoded = self.encoder.transform(x[ohe_cols])
        encoded_df = pd.DataFrame(
            encoded,
            columns=self.encoder.get_feature_names_out(ohe_cols),
            index=x.index
        )
        x = x.drop(ohe_cols, axis=1)
        x = pd.concat([x, encoded_df], axis=1)
        return x
```

MISSINF CATEGORICAL 

```python
cols4 = [
    'BsmtFinType1',
    'BsmtFinType2',
    'BsmtExposure',
    'BsmtCond',
    'BsmtQual',
    'GarageType',
    'GarageFinish',
    'GarageQual',
    'GarageCond',
    'KitchenQual'
]
class MissingImputer(BaseEstimator, TransformerMixin):
    def fit(self, x, y=None):
        return self
    def transform(self, x):
        x = x.copy()
        x[cols4] = x[cols4].fillna('Missing')
        return x
```

LOG TRANSFORMATION

```python
cols5 = [
    'MSSubClass', 'LotArea', 'OverallQual', 'OverallCond',
    'YearBuilt', 'YearRemodAdd', 'MasVnrArea', 'BsmtFinSF1',
    'BsmtFinSF2', 'BsmtUnfSF', 'TotalBsmtSF', '1stFlrSF',
    '2ndFlrSF', 'GrLivArea', 'BsmtFullBath',
    'BsmtHalfBath', 'FullBath', 'HalfBath', 'BedroomAbvGr',
    'KitchenAbvGr', 'TotRmsAbvGrd', 'Fireplaces',
    'GarageYrBlt', 'GarageCars', 'GarageArea', 'WoodDeckSF',
    'OpenPorchSF', 'EnclosedPorch',
    'ScreenPorch', 'MoSold', 'YrSold'
]
class LogTransformer(BaseEstimator, TransformerMixin):
    def fit(self, x, y=None):
        return self
    def transform(self, x):
        x = x.copy()
        x[cols5] = np.log1p(x[cols5])
        return x
```

```python
x_train.hist(figsize=(30,30))
```

```python
cat_cols = [
    'MSZoning',
    'Utilities',
    'Exterior1st',
    'Exterior2nd',
    'Functional',
    'SaleType'
]
class CatImputer(BaseEstimator, TransformerMixin):
    def fit(self, x, y=None):
        self.si = SimpleImputer(strategy='most_frequent')
        self.si.fit(x[cat_cols])
        return self
    def transform(self, x):
        x = x.copy()
        x[cat_cols] = self.si.transform(x[cat_cols])
        return x
```

```python
param_grid = {
    'model__alpha': [0.0001, 0.001, 0.01, 0.1, 1]
}

grid = GridSearchCV(
    estimator=pipe,
    param_grid=param_grid,
    cv=5,
    scoring='r2',
    n_jobs=-1
)

grid.fit(x_train, y_train)
```

```python
pipe = Pipeline([
    ('drop_cols', dropper()),
    ('missing_impute', MissingImputer()),
    ('num_impute', numimputer()),
    ('cat_impute',CatImputer()),
    ('ordinal_encode', ordinalencoder()),
    ('log_transform', LogTransformer()),
    ('one_hot_encode', OHEEncoder()),
    ('scaler', StandardScaler()),
    ('model', Lasso(alpha=1))
])
```

```python
pipe.fit(x_train, y_train)
```

```python
x = dropper().fit_transform(x_train)
print("dropper done")

x = MissingImputer().fit_transform(x)
print("missing done")

x = numimputer().fit_transform(x)
print("num done")

x = ordinalencoder().fit_transform(x)
print("ordinal done")

x = LogTransformer().fit_transform(x)
print("log done")

x = OHEEncoder().fit_transform(x)
print("ohe done")
```

```python
y_pred = pipe.predict(x_test)
```

```python
print("R2:", r2_score(y_test, y_pred))
print("RMSE:", np.sqrt(mean_squared_error(y_test, y_pred)))
```
